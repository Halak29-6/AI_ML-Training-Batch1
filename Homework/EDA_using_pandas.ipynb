`Exploratory Data analysis (EDA) using Pandas`

```python
import pandas as pd
import numpy as np
```

```python
data = {
    'Employee_ID': [101, 102, 103, 104, 105, 106, 107, 104],
    'Age': [25, np.nan, 30, 45, 30, 22, 50, 45],
    'Salary': [50000, 60000, 55000, 80000, 55000, np.nan, 90000, 80000],
    'Department': ['IT', 'HR', 'IT', 'Finance', 'IT', 'HR', 'Finance', 'Finance'],
    'Name': ['Halak', 'Smith', 'Michael ', 'abhishek', 'Riya' , 'Emily' , 'Daniel','george']
}
df = pd.DataFrame(data)
```

```python
print(df.head())
```

```python
print(f"\nDataset Shape: {df.shape}")
```

```python
df.info()
```

```python
print(df.describe().T)
```

```python
df.isnull().sum()
```

```python
print(df.isnull().sum())
```

```python

duplicate_count = df.duplicated().sum()
```

```python

df_cleaned = df.drop_duplicates()
df_cleaned['Age'] = df_cleaned['Age'].fillna(df_cleaned['Age'].median())
```

```python

print(df_cleaned['Department'].value_counts())
```

```python
avg_salary_by_dept = df_cleaned.groupby('Department')['Salary'].mean()
print("Average Salary by Department:")
print(avg_salary_by_dept)
```

```python

print(df_cleaned[['Age', 'Salary']].corr())
```
