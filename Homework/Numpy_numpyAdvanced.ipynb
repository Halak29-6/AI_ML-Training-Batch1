```python
nums2d = np.array([[-9,8,-7],[6,-5,4]])
print("initial array: \n",nums2d)
#normal broadcasting

pos_nums = nums2d[nums2d>0]
print("\npositive nums:" , pos_nums)

neg_nums = nums2d[nums2d<0]
print("\nnegative nums: ",neg_nums)

print("------------\n------------")

# using masking
print("using masking")
pos_mask = nums2d>0
neg_mask = nums2d<0

pos_nums = nums2d[pos_mask]
neg_nums = nums2d[neg_mask]

print(pos_nums)
print(neg_nums)
```
