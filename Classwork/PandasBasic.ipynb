# Merg

```python
df1 = pd.DataFrame({ 'ID' : [1,2,34,4],
                     'Class' : [5,6,7,8]})
df1
```

```python
df2 = pd.DataFrame({ 'ID' : [1,2,3,4],
                     "Name" : ['A','B','C','D']})
df2
```

```python
pd.merge(df1,df2)
```

```python
pd.merge(df1,df2, on = "ID")
```

```python
pd.merge(df2,df1, on = "ID")
```

```python
# intersection = inner , same as merge on id(same valuse merge different value not)
pd.merge(df1,df2, on = 'ID' , how = 'inner')
```

# Concat()

padas provides various facilities for easily combining together Series, DataFrame and panel objects

```python
sr1 = pd.Series([0,1,2])
sr1
```

```python
df2 = pd.DataFrame({ "ID" : [5,6,7,8],
                     'Name' : ['E','F','G','H'],
                     'Class' : [9,10,11,12]})
df2
```

```python
pd.concat([df1,df2])
```

```python
pd.concat([df1,df2], axis = 1)
```

```python
pd.concat([df2,df1], axis = 0)
```

```python
pd.concat([df1,df2], axis = 0, ignore_index = True)
```
