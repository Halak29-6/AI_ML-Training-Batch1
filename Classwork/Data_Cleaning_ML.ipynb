***Handling missing values***

```python
import numpy as np
```

```python
data = {
    'Age': [23,np.nan,20,18,np.nan],
    'Score':[56,87,23,np.nan,57],
    'City':['ny','paris','london',np.nan,'rome']
}
dataf= pd.DataFrame(data)
print("Original data:\n\n",dataf)
# is null , not null , nan ,
```

```python
print(dataf.isnull())
```

```python
print(dataf.isnull().sum())
```

```python
print(dataf.isnull().sum().sum())
```

```python
print(dataf.notnull())
```

***Delete missing values***

```python
df_clean_rows = dataf.dropna()
df_clean_cols = dataf.dropna(axis=1)
df_clean_subset = dataf.dropna(subset=['Age'])
```

```python
df_clean_rows
```

```python
df_clean_cols
```

```python
df_clean_subset
```

```python
df_zero = dataf.fillna(0)
```

```python
dataf['Age'] = dataf['Age'].fillna(dataf['Age'].mean())
```

```python
dataf['City'] = dataf['City'].fillna(dataf['City'].mode()[0])
```

```python
df_zero
```

```python
dataf['Age']
```

```python
dataf['City']
```

```python
dataf['Score']
```
